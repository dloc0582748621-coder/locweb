## 1. Jupyter Notebook vs JupyterLab

### Jupyter Notebook (Classic)
- Giao dien don gian, de dung
- 1 file = 1 tab trinh duyet
- Phu hop: hoc, thuc hanh nhanh
- Khoi dong: `jupyter notebook`

### JupyterLab (Modern)
- Giao dien IDE day du
- Nhieu tab, keo tha file, terminal tich hop
- File explorer, variable inspector
- Phu hop: du an lon, lam viec chuyen nghiep
- Khoi dong: `jupyter lab`

**Khoi tao:** Doi voi khoa hoc nay, su dung **JupyterLab** la khuyen nghi.

## 3. Phim tat quan trong

| Phim | Hanh dong |

## 4. Magic Commands

**Magic commands** la cac lenh dac biet cua IPython (kernel cua Jupyter), bat dau bang `%` (line magic) hoac `%%` (cell magic).

### Line magics (%):
- `%timeit` -- do thoi gian thuc hien 1 dong lenh
- `%run` -- chay mot file .py
- `%who` / `%whos` -- liet ke bien trong session
- `%pwd` -- thu muc hien tai
- `%ls` -- liet ke file trong thu muc
- `%matplotlib inline` -- hien thi do thi trong notebook
- `%load_ext` -- tai extension
- `%reset` -- xoa het bien trong session

### Cell magics (%%):
- `%%timeit` -- do thoi gian ca cell
- `%%writefile` -- luu cell vao file
- `%%bash` -- chay lenh bash trong cell
- `%%html` -- render HTML trong cell


```python
# =============================================
# Magic command: %timeit
# Do thoi gian thuc hien mot phep tinh
# =============================================

# So sanh 2 cach tao danh sach

# Cach 1: Vong lap for thuong
%timeit result = [i**2 for i in range(1000)]

# Cach 2: List comprehension (thuc ra giong nhau, vi du minh hoa)
%timeit result = list(map(lambda i: i**2, range(1000)))
```

```python
# =============================================
# Magic command: %%timeit (Cell magic)
# Do thoi gian TOAN BO cell
# =============================================

%%timeit
# Tinh tong cac so tu 1 den 10000
tong = 0
for i in range(10000):
    tong += i
```

```python
# =============================================
# Magic command: %matplotlib inline
# Ve do thi trong notebook
# =============================================

%matplotlib inline
import matplotlib.pyplot as plt
import math

# Tao du lieu
x = [i * 0.1 for i in range(0, 63)]  # 0 den 6.2 (xap xi 2*pi)
y_sin = [math.sin(v) for v in x]
y_cos = [math.cos(v) for v in x]

# Ve do thi
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(x, y_sin, color='blue', linewidth=2, label='sin(x)')
axes[0].set_title('Ham sin(x)')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
axes[0].axhline(y=0, color='black', linewidth=0.5)

axes[1].plot(x, y_cos, color='red', linewidth=2, label='cos(x)')
axes[1].set_title('Ham cos(x)')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].legend()
axes[1].grid(True, alpha=0.3)
axes[1].axhline(y=0, color='black', linewidth=0.5)

plt.suptitle('Do thi ham luong giac -- Module 02 Demo', fontsize=13)
plt.tight_layout()
plt.show()
print('Do thi da hien thi trong notebook!')
```

## 5. Xuat Notebook ra cac dinh dang

Jupyter ho tro xuat notebook ra nhieu dinh dang khac nhau:

| Dinh dang | Lenh | Mo ta |
|-----------|------|-------|
| HTML | `jupyter nbconvert --to html notebook.ipynb` | Trang web, de chia se |
| PDF | `jupyter nbconvert --to pdf notebook.ipynb` | Can cai LaTeX |
| Script Python | `jupyter nbconvert --to script notebook.ipynb` | Chuyen sang .py |
| Slides | `jupyter nbconvert --to slides notebook.ipynb` | Dung reveal.js |
| Markdown | `jupyter nbconvert --to markdown notebook.ipynb` | Van ban Markdown |

**Cach xuat trong giao dien:**
- JupyterLab: `File` -> `Save and Export Notebook As...`
- Classic Notebook: `File` -> `Download as`

### Xuat bang lenh (!shell command trong cell):


```python
# Xuat notebook hien tai sang HTML
# Dau ! dung de chay lenh shell tu ben trong Jupyter

# Xem ten file notebook hien tai
import os
import json

# Lay ten notebook hien tai (JupyterLab)
try:
    notebook_name = [f for f in os.listdir('.') if f.endswith('.ipynb')]
    print('Cac file notebook trong thu muc hien tai:')
    for nb in notebook_name:
        size_kb = os.path.getsize(nb) / 1024
        print(f'  {nb} ({size_kb:.1f} KB)')
except Exception as e:
    print(f'Khong the liet ke: {e}')

print('\nDe xuat sang HTML, chay trong terminal:')
print('  jupyter nbconvert --to html 02_jupyter_notebook.ipynb')
print('\nDe xuat sang Python script:')
print('  jupyter nbconvert --to script 02_jupyter_notebook.ipynb')
```
